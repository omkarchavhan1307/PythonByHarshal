1. Write a program to find the union of two sets without using the union() method.

```python
set1 = {1,2,3,4,5,6,7,8}
set2 = {6,7,8,9,10}

union = set1.copy()

for i in set2:
    union.add(i)

print("Union of two sets is :",union)
```

2. Write a program to remove all duplicate elements from a list using a set. 

```python
listone = [1,2,3,4,5,6,5,4,3,2,7,8,9,10]

unique = list(set(listone))

print("List after removing duolicates :",unique)
```

3. Write a program to find all common elements between two lists using sets. 

```python
ListOne = [1,2,3,4,5,6,7]
ListTwo = [5,6,7,8,9,10]

SetOne = set(ListOne)
SetTwo = set(ListTwo)

common = SetOne.intersection(SetTwo)

print("Common elements are :",common)
```

4. Write a program to count the number of unique elements in a list using a set. 

```python
numbers = [1,2,3,4,5,6,5,4,3,7,8,9,10,11,12]

unique = set(numbers)

print("Unique numbers :",list(unique))
print("Count of numbers :",len(unique))
```

6. Write a program to create a new set containing only the even numbers from a given set. 

```python
numbers = {1,2,4,5,7,8,14,17,20,22,50,101}

even = set()

for i in numbers:
    if i % 2 == 0:
        even.add(i)

print("Even numbers from given set are :",even)
```

7. Write a program to merge three sets into a single set containing only unique elements. 

```python
set1 = {1,2,3,4,5,6,7,8}
set2 = {5,6,7,8,9,10,11}
set3 = {9,10,11,12,13,14}

merge = set1.copy()

for i in set2:
    merge.add(i)

for i in set3:
    merge.add(i)

print("Unique elements after merging 3 sets :",merge)
```

8. Given two strings, find the common characters using sets. 

```python
str1 = "Omkar"
str2 = "Komal"

set1 = set(str1.upper())
set2 = set(str2.upper())

common = set1.intersection(set2)

print("Common characters are :",common)
```
