```python
class Engine:
    def __init__(self, engine_no, cc):
        self.engine_no = engine_no
        self.cc = cc

    def start(self):
        print("Engine started..!")

class Car:
    def __init__(self, model, price, engine):
        self.model = model
        self.price = price
        self.engine = engine

    def display(self):
        print("Engine no. :",self.engine.engine_no)
        print("Engine's cc :",self.engine.cc)
        print("Model of Car :",self.model)
        print("Price of Car :",self.price)

    def start_car(self):
        self.engine.start()


engine = Engine("ENG12345", 600)

obj1 = Car("Scorpio", 900000, engine)

obj1.display()
obj1.start_car()
```

3. Bank Account — Encapsulation Create a BankAccount class. • Create private variable __balance.  • Create deposit() and withdraw() methods.  • Create get_balance() getter.  • Don't allow withdrawal greater than balance.  • Don't allow negative deposit.  • Keep balance inaccessible directly from outside.  

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
        print("Balance at the start :",self.__balance)
        print(" ")

    def deposit(self, ammount):
        self.ammount = ammount
        if ammount > 0:
            self.__balance += ammount
            print("Deposiited ammount :",self.ammount)
            print("Ammount deposited successfully..!")
            print("Balance after deposited :",self.__balance)
            print(" ")

        else:
            print("Invalid Ammount.")

    def withdraw(self, ammount):
        self.ammount = ammount
        if ammount <= self.__balance:
            self.__balance -= self.ammount
            print("Withdraw ammount :",self.ammount)
            print("Ammount withdraw successfully..!")
            print("Balance after withdraw :",self.__balance)
            print(" ")

        else:
            print("Insufficient Balance..")

    def get_balance(self):
        return self.__balance


obj1 = BankAccount(50000)

obj1.deposit(10000)
obj1.withdraw(30000)
```

4. Student Marks — Encapsulation + Getter/Setter Create a Student class. • Private variable __marks.  • Create getter and setter.  • Setter should allow marks only between 0 and 100.  • Create display() method.  • Try changing marks with both valid and invalid values.  

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        self.__marks = marks
        if marks >= 0:
            self.__marks += marks
            print("Marks updated successfully...!")

        else:
            print("Please enter marks in between 0 to 100.")

    def display(self):
        print("Student's marks :",self.__marks)
        print(" ")


obj1 = Student(95)

obj1.display()
obj1.set_marks(98)

```
