2. Take the length and width of a rectangle as input and calculate: -Area  - Perimeter 

```python
length = int(input("Enter length of rectangle :"))
breadth = int(input("Enter breadth of rectangle :"))

area = length * breadth
perimeter = ((2 * length) + (2 * breadth))

print("Area of rectangle is :",area)
print("Perimeter of rectangle is :",perimeter)
```

3. Take the price and quantity of a product from the user and calculate: - Total Amount  - GST (18%) - Final Bill 

```python
product_name = input("Enter product name :")
price = float(input("Price of product :"))
quantity = int(input("Enter quantity of product :"))

total_amount = price * quantity
gst = (total_amount * 18) / 100
final_bill = total_amount + gst

print("Product Name :",product_name)
print("Total amount of product :",total_amount)
print("GST on product :",gst)
print("Final bill of product :",final_bill)
```

7. Write a program to determine whether a year is a Leap Year.

```python
year = int(input("Enter Year :"))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

8. Create a simple calculator using match-case.

- Operations:

• Addition 
• Subtraction 
• Multiplication 
• Division

```python
a = int(input("Enter First Number :"))
b = int(input("Enter Second Number :"))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter Choice(1/2/3/4) :"))

match choice:
    case 1:
        print("Addition is =", a + b)
    case 2:
        print("Subtraction is =", a - b)
    case 3:
        print("Multiplication is =", a * b)
    case 4:
        if b != 0:
            print("Division is =", a / b)
        else:
            print("Division by Zero is not allowed")
    case _:
        print("Invalid Choice")
```
