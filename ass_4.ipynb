1. Write a program to count the number of vowels and consonants in a given string. 

```python
string = input("Enter your string :").lower()

v = 0
c = 0

for i in string:
    if i in "aeiou":
        v += 1
    elif i.isalpha():
        c += 1

print("Vowels :",v)
print("Consonants :",c)
```

2. Write a program to check whether a given string is a Palindrome. -Input: madam ,-Output : Palindrome 

```python
string = input("Enter your string :")

temp = "".join(reversed(string))

if string == temp :
    print("It is Palindrome.")
else:
    print("It is not Palindrome.")
```

3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions. 

```python
string = "Omkar"      # input("Enter your name :")

reversed = ""

for i in string:
    reversed = i + reversed

print("Reversed string is :",reversed)
```

5. Write a program to remove all duplicate characters from a string while maintaining their original order. -Input : programming -Output : progamin 

```python
string = "programming"

result = ""

for i in string:
    if i not in result:     # This condition is false when duplicate is there.
        result += i

print("Output :", result)
```

6. Write a program to find the longest word in a sentence. -Input: Python is an amazing programming language  -Output: programming 

```python
string = " Python is an amazing programming language."

words = string.split()       # ["Python", "is", "an", "amazing", "programming", "language"]     
longest = ""

for i in words:
    if len(i) > len(longest):   # If the current word is longer than the previous longest word, update longest.
        longest = i
print("Longest word is :",longest)

```

7. Write a program to capitalize the first letter of every word without using the title() method.  -Input:welcome to python -Output:   Welcome To Python 

```python
string = "welcome to python"

words = string.split()

for i in words:
    print(i[0].upper() + i[1:], end=" ")
```

8. Write a program to count the total number of words in a sentence without using the split() method. -Input:Python is easy to learn  -Output:5 

```python
sentence =  "Python is easy to learn" # input("Enter a sentence: ")

count = 1

for i in sentence:
    if i == " ":
        count += 1

print("Total Words =", count)
```

9. Write a program to remove all spaces from a string. - Python Programming Language 

```python
string = "Python Programming Language" # input("Enter a string: ")

result = ""

for i in string:
    if i != " ":
        result += i

print("String without spaces:", result)
```
