2. Write a Python program using nested loops to find and print all prime numbers between 1 and 100.  

```python
for num in range(2,101):
    is_prime = True
    for i in range(2,num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")
```

3. Given two lists, use nested loops to find and print the common elements:  list1 = [10, 20, 30, 40, 50] list2 = [20, 40, 60, 80, 100] 

```python
List1 = [10,20,30,40,50]
List2 = [20,40,60,80,100]

for i in List1:
    for j in List2:
        if i == j:
            print(i)
```

4. Write a Python program to count the frequency of each character in a string without using count().  

```python
String = "My name is Omkar"

frequency = {}

for i in String:
    if i in frequency:
        frequency[i] += 1

    else:
        frequency[i] = 1

print(frequency)
```

5. Write a Python program to find the first non-repeated character in a string.  

```python
String = "aabbccdee"

frequency = {}

for i in String:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

for i in String:
    if frequency[i] == 1:
        print("Non-repeated character is :",i)
        break
```
