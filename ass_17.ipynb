3. Bank Account — Encapsulation Create a BankAccount class. • Create private variable __balance.  • Create deposit() and withdraw() methods.  • Create get_balance() getter.  • Don't allow withdrawal greater than balance.  • Don't allow negative deposit.  • Keep balance inaccessible directly from outside.  

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
        print("Balance at the start :",self.__balance)
        print(" ")

    def deposit(self, ammount):
        self.ammount = ammount
        if ammount > 0:
            self.__balance += ammount
            print("Deposiited ammount :",self.ammount)
            print("Ammount deposited successfully..!")
            print("Balance after deposited :",self.__balance)
            print(" ")

        else:
            print("Invalid Ammount.")

    def withdraw(self, ammount):
        self.ammount = ammount
        if ammount <= self.__balance:
            self.__balance -= self.ammount
            print("Withdraw ammount :",self.ammount)
            print("Ammount withdraw successfully..!")
            print("Balance after withdraw :",self.__balance)
            print(" ")

        else:
            print("Insufficient Balance..")

    def get_balance(self):
        return self.__balance


obj1 = BankAccount(50000)

obj1.deposit(10000)
obj1.withdraw(30000)
```

8. Vehicle — Method Overriding Create a parent class Vehicle. • Create a method start().  • Create child classes Car, Bike, and Bus.  • Override start() in each child.  • Create objects of all three classes.  • Call start() and observe different outputs.  

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting....")

class Car(Vehicle):
    def start(self):
        print("Car satrted...")

class Bike(Vehicle):
    def start(self):
        print("Bike started...")

class Bus(Vehicle):
    def start(self):
        print("Bus started...")

obj1 = Car()
obj2 = Bike()
obj3 = Bus()

obj1.start()
obj2.start()
obj3.start()

# the child classes overrides the parent's start() method.
```

9. Payment System — Method Overriding + Polymorphism Create a parent class Payment. Create child classes: Payment ├── UPI ├── CreditCard └── NetBanking • Create pay() method in parent.  • Override pay() in each child.  • Create a function make_payment(payment).  • Pass different payment objects to it.  • Observe different behavior.  

```python
class Payment:
    def pay(self):
        print("Pay via UPI, Credit Card and Netbanking.")

class Upi(Payment):
    def pay(self):
        print("Pay using UPI.")

class Creditcard(Payment):
    def pay(self):
        print("Pay using Credit card.")

class Netbanking(Payment):
    def pay(self):
        print("Pay using Netbanking.")

def make_payment(Payment):
    Payment.pay()

obj1 = Upi()
obj2 = Creditcard()
obj3 = Netbanking()

make_payment(obj1)
make_payment(obj2)
make_payment(obj3)
```

```python
# class Calculator:
#     def add(self, num1 = 0, num2 = 0, num3 = 0, num4 = 0):
#         print(num1 + num2 + num3 + num4)

# obj1 = Calculator()

# obj1.add(10,20)
# obj1.add(10,20,30)
# obj1.add(10,20,30,40)

class Calculator:
    def add(self, *args):
        return sum(args)

    def multiply(self, *args):
        result = 1

        for i in args:
            result = result * i

        return result

obj1 = Calculator()

print("Addition method")
print(obj1.add(10,20))
print(obj1.add(10,20,30))
print(obj1.add(10,20,30,40))
print(obj1.add(10,20,30,40,50))
print(" ")

print("Multiply method")
print(obj1.multiply(10,20))
print(obj1.multiply(10,20,30))
print(obj1.multiply(10,20,30,40))
print(obj1.multiply(10,20,30,40,50))
```

11. Library Management — Aggregation + Encapsulation Create: Book Library • Book should contain title, author, and price.  • Make price private.  • Create getter/setter for price.  • Library should contain multiple Book objects.  • Add books to the library.  • Display all books.  • Search for a book by title.  

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if self.__price > 0:
            self.__price = price
            self.get_price()
            print("After update price is :",self.__price)
            print(" ")

        else:
            print("Invalid ammount.")

    
    def display(self):
        print("Title :", self.title)
        print("Author :", self.author)
        print("Price :", self.__price)
        print(" ")

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def display_book(self):
        for book in self.books:
            book.display()

    def search_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                print("Book found :")
                print(" ")
                book.display()
                return

        print("Book not found.")
    
Book1 = Book("Python", "John", 500)
Book2 = Book("Java", "James", 600)
Book3 = Book("SQL", "David", 400)

Book1.display()
Book2.display()
Book3.display()

Book1.set_price(300)
Book1.display()

Library1 = Library()

Library1.add_book(Book1)
Library1.add_book(Book2)
Library1.add_book(Book3)


print("Books in Library :")
print(" ")
Library1.display_book()
Library1.search_book("Python")
```

12. E-Commerce Product — Encapsulation + Static Method Create a Product class. • Private variable __price.  • Instance variables: name, quantity.  • Getter/setter for price.  • Instance method total_price().  • Static method calculate_gst(price) that calculates GST.  • Display the final product price including GST

```python
class Product:
    def __init__(self, name, quantity, price):
        self.name = name
        self.quantity = quantity
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        self.__price = price
        if self.__price > 0:
            self.__price = price
            print("After updating price :",self.__price)
            print(" ")
        else:
            print("Invalid ammount.")

    def total_price(self):
        return self.__price * self.quantity


    @staticmethod
    def calculate_gst(price):
        return price * 0.18

    def display(self):
        total = self.total_price()
        gst = self.calculate_gst(total)

        final_price = total + gst

        print("Product name :",self.name)
        print("Quantity :",self.quantity)
        print("Price of product :",self.get_price())
        print("Total price without gst :",self.total_price())
        print("GST :",gst)
        print("Final price is :", final_price)
        print(" ")

Product1 = Product("Laptop", 3, 50000)

Product1.display()

Product1.set_price(40000)

Product1.display()
```

Hospital Management — Aggregation + Inheritance Create: Person ├── Doctor └── Patient and an Address class. • Doctor and Patient should inherit from Person.  • Both should contain an Address object.  • Add different attributes for Doctor and Patient.  • Override display() in both child classes.  • Use super() where required.  This combines inheritance + aggregation + overriding. 

```python
class Address:
    def __init__(self, city, state, pincode):
        self.city = city
        self.state = state
        self.pincode = pincode

    def display_address(self):
        print("City :",self.city)
        print("State :",self.state)
        print("Pincode :",self.pincode)

class Person:
    def __init__(self, name, age, address):
        self.name = name
        self.age = age
        self.address = address

    def display_person(self):
        print("Person name :",self.name)
        print("Person age :",self.age)
        self.address.display_address()

class Doctor(Person):
    def __init__(self, name, age, address, specialization):
        super().__init__(name, age, address)
        self.specialization = specialization

    def display_doctor(self):
        print("Doctor's Information :",self.display_person())
        print("Specialization in :",self.specialization)

class Patient(Person):
    def __init__(self, name, age, address, desease):
        super().__init__(name, age, address)
        self.desease = desease

    def display(self):
        print("Patient's Information",self.display_person())
        print("Desease :",self.desease)

Address1 = Address("Pune", "Maharashtra", 412406)
Address2 = Address("Mumbai", "Maharashtra", 420406)
Doctor1 = Doctor(
    "Dr. Sharma",
    40,
    Address1,
    "Cardiologist"
)

Patient1 = Patient(
    "Rahul",
    25,
    Address2,
    "Fever"
)

print("Doctor Details :")
print(" ")
Doctor1.display_doctor()
print(" ")
print("--------------------------------")
print(" ")
print("Patient Details :")
print(" ")
Patient1.display()




```
