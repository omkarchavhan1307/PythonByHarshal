```python
str1 = input("Enter a sentence :").lower()

v = 0
c = 0
s = 0

for i in str1:
    if i in "aeiou":
        v += 1
    elif i.isalpha():
        c += 1
    elif i in " ":
        s += 1

print("Vowels :",v)
print("Consonants :",c)
print("Spaces :",s)
```

```python
def top(marks):
    highest_marks = max(marks.values())

    for name in marks:
        if marks[name] == highest_marks:
            print("Topper Name:", name)
            print("Highest Marks:", highest_marks)

top({"Omkar": 90, "Komal": 85, "Nikhil": 95})
```
