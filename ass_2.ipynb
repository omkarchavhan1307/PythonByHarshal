1. Create variables of all Python primitive data types. Print their:
• Value 
• Data Type  
• Memory Address (id()) 

```python
# Integer
num1 = 10
print(type(num1))
print(id(num1))

# Float
percentage = 67.83
print(type(percentage))
print(id(percentage))

# Complex
comp_num = 13+7j
print(type(comp_num))
print(id(comp_num))

# Boolean
isStudent_Present = True 
print(type(isStudent_Present))
print(id(isStudent_Present))

# String
student_name = "Omkar Santosh Chavhan"
print(type(student_name))
print(id(student_name))

# NoneType
hobby = None
print(type(hobby))
print(id(hobby))

```

5. Write a program to calculate electricity bill.
 Use variables for: • Units Consumed  • Cost per Unit  • Calculate total bill and GST (18%). 

```python
unit_consumed = 280
cost_per_unit = 12

total_bill = unit_consumed + cost_per_unit
gst = total_bill * (18/100)
final_bill = total_bill + gst

print("Units consumed by customer :",unit_consumed)
print("Cost of single unit :",cost_per_unit)
print("Total bill without GST :",total_bill)
print("Total GST :",gst)
print("Final bill of customer including GST :",final_bill)
```

10. Create a program that stores product information. 

Store: • Product Name  • Price  • Quantity  

Calculate: • Total Cost  • GST (18%)  • Final Amount 

```python
product_name = "Laptop"
price = 60000
quantity = 4

total_cost = price * quantity
gst = total_cost * 0.18
final_amount = total_cost + gst

print("Product Name :", product_name)
print("Price        :", price)
print("Quantity     :", quantity)
print("Total Cost   :", total_cost)
print("GST (18%)    :", gst)
print("Final Amount :", final_amount)
```
