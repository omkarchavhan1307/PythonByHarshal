```python
# Zero Division Error

a = 10
b = 0
print(a/b)
```

```python
a = 10
b = 0

try:
    num = a/b

except ZeroDivisionError as err:
    print("You cannot divide by zero.",err)

else:
    print("Division is :",num)
```

```python
# Value Error

try:
    a = int(input("Enter num1 :"))
    b = int(input("Enter num2 :"))

    res = a + b

except ValueError as err:
    print("You enter invalid value.",err)

else:
    print("Addition is :",res)
```

```python
# Type error

try:
    a = int(input("Enter first number :"))
    b = input("Enter second number :")

    res = a + b

except TypeError as err:
    print("You enter wrong data type.",err)

else:
    print("Addition is :",res)
```

```python
# Index Error
ListOne = [10,20,30,40,50]

try:
    temp = ListOne[5]

except IndexError as err:
    print("Index is not found.",err)

else:
    print(temp)
```

```python
# Key Error

DictOne = {"name" : "Omkar", "age" : "22", "roll_no" : "08"}

try:
    user = DictOne["marks"]

except KeyError as err:
    print("Key is not found.",err)

else:
    print(user)
```

```python
# Name Error

num1 = 10

try:
    res = num1 + num5

except NameError as err:
    print("Value is not defined.",err)

else:
    print(res)
```

```python
# Attribute Error

ListOne = [10,20,30,40,50,60]

try:
    ListOne.add[70]

except AttributeError as err:
    print("Add method is not available.",err)

else:
    print(ListOne)
```

```python
# File Error

try:
    file_data = open("temp.txt")

    data = file_data.read()

except FileNotFoundError as err:
    print("File is not found.",err)

else:
    print(data)
```

```python
# Import Error

try:
    from math import xyz
    res = xyz(25)

except ImportError as err:
    print("This method is not available in Python.",err)

else:
    print(res)
```

```python
# Module not found

try:
    import demo

except ModuleNotFoundError as err:
    print("Given module is not found.",err)

else:
    print(demo)
```

```python
# Generic Exception Handling

a = 10
b = 0

try:
    res = a/b

except Exception as err:
    print("Not divided by 0.",err)

else:
    print(res)
```
