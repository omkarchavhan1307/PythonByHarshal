1. Student Class – __str__() Create a Student class with name, roll_no, and marks. Implement __str__() so that printing the name, roll_no, and marks.

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def __str__(self):
        return f"Name : {self.name}, Roll no. : {self.roll_no}, Marks : {self.marks}"

obj1 = Student("Omkar", 101, 70)

print(obj1)
        
```

2. Product Class – __eq__() Create a Product class with name and price. Implement __eq__() to check whether two products have the same price. 

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(value1, value2):
        return value1.price == value2.price


obj1 = Product("Mobile", 20000)
obj2 = Product("Headphone", 20000)

print(obj1 == obj2)
```

3. Shopping Cart – __len__() Create a ShoppingCart class containing a list of products. Implement __len__() so that len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def __len__(self):
        return len(self.products)


obj1 = ShoppingCart()

obj1.add_product("Laptop")
obj1.add_product("Mobile")
obj1.add_product("I-phone")
obj1.add_product("TV")

print(len(obj1))
```

4. Bank Account – __str__() Create a BankAccount class with account holder name, account number, and balance. Implement __str__() to display account information when the object is printed. 

```python
class BankAccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance

    def __str__(self):
        return f"Name : {self.name}, Account_No : {self.account_no}, Balance : {self.balance}"


obj1 = BankAccount("Omkar", 10001, 50000)

print(obj1)
```

5. Number Class – __add__() Create a Number class that stores one number. Implement __add__() so that two objects can be added.

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

obj1 = Number(40)
obj2 = Number(50)

print(obj1 + obj2)
```

6. Create a Number class and implement: -__add__()  -__sub__()  -__mul__()  -__truediv__()  -__mod__() 

```python
class Number:
    def __init__(self, num1):
        self.num1 = num1

    def __add__(self, other):
        return self.num1 + other.num1

    def __sub__(self, other):
        return self.num1 - other.num1

    def __mul__(self, other):
            return self.num1 * other.num1

    def __truediv__(self, other):
            return self.num1 / other.num1

    def __mod__(self, other):
            return self.num1 % other.num1


obj1 = Number(50)
obj2 = Number(40)

print(obj1 + obj2)
print(obj1 - obj2)
print(obj1 * obj2)
print(obj1 / obj2)
print(obj1 % obj2)
```
