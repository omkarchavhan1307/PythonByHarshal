1. Write a function that accepts two numbers and returns the sum, difference, using multiple return values. 

```python
def calculate(a,b):                                   # create a calculate function in which accept the two numbers a 
                                                      #   and b.
    return(a + b , a - b)                             # returns the value of a+b and a-b in calculate.

num1 = int(input("Enter a first number :"))           # Take input from user.
num2 = int(input("Enter a second number :"))

sum_result, difference_result = calculate(num1, num2) # calculated values are stored in sum_result, difference_result hence we have to print these two.

print("Sum of two numbers :",sum_result)              # print the result
print("Difference of two numbers :",difference_result)
```

2. Write a function that accepts a list of numbers and returns the largest and smallest number. 

```python
def largest_smallest(numbers):          # create a function largest_smallest with parameter numbers.
    largest = numbers[0]                # create largest variable with 0 value from numbers.
    smallest = numbers[0]               # create smallest variable with 0 value from numbers.

    for num in numbers:                 # iterate values from numbers
        if num > largest:               # if num is greater than value in largest insert num into largest.
            largest = num
        if num < smallest:              #  # if num is smaller than value in smallest insert num into smallest.
            smallest = num

    return largest, smallest            # return both values into function.

numbers = [10, 25, 5, 40, 18]

large, small = largest_smallest(numbers)  # store output in large and small

print("Largest Number =", large)
print("Smallest Number =", small)
```

3. Write a function using  *args to calculate the sum, average, maximum, and minimum of any number of integers.  

```python
def statistics(*args):               # create function statistics
    total = sum(args)                # sum of all values in total
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)

    return total, average, maximum, minimum    # return all the results into statistics.

total, avg, maximum, minimum = statistics(10, 20, 30, 40, 50)  # Pass the arguments and store the result in variables positionally.

print("Sum :", total)
print("Average :", avg)
print("Maximum :", maximum)
print("Minimum :", minimum)
```

5. Write a function that accepts a string and returns the number of uppercase letters, lowercase letters, digits. 

```python
def count_characters(string):
    upper = 0
    lower = 0
    digit = 0

    for ch in string:
        if ch.isupper():
            upper += 1
        if ch.islower():
            lower += 1
        if ch.isdigit():
            digit += 1

    return(upper, lower, digit)

string = input("Enter a string :")

upper,lower,digit = count_characters(string)
print(string)
print("Numbers of uppercase :",upper)
print("Numbers of Lowercase :",lower)
print("Numbers of Digit :",digit)


   
```

6. Write a function that checks whether a given string is a Palindrome.  

```python
def palindrome(string):
    if string == string[::-1]:
        return True
    else:
        return False

string = input("Enter a string :")

print(string)

if palindrome(string):
    print("String is Palindrome.")
else:
    print("STring is not Palindrome.")
    
```

7. Write a function that accepts a list of integers and returns a new list containing only even numbers. 

```python
def even_numbers(lst):
    print("Even Numbers are:")
    for num in lst:
        if num % 2 == 0:
            print(num, end=" ")

numbers = [12, 7, 18, 25, 30]

even_numbers(numbers)
```

8. Create a menu-driven calculator using functions that performs Addition, Subtraction, Multiplication, Division, and Exit.  

```python
def calculator():
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter your choice: "))

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print("Answer =", a + b)

    elif choice == 2:
        print("Answer =", a - b)

    elif choice == 3:
        print("Answer =", a * b)

    elif choice == 4:
        if b != 0:
            print("Answer =", a / b)
        else:
            print("Division by zero is not possible.")

    else:
        print("Invalid Choice")

calculator()
```

9. Write a function that accepts a dictionary of students and their marks, then returns the name of the topper and the highest marks.

```python
def top(marks):
    highest_marks = max(marks.values())

    for name in marks:
        if marks[name] == highest_marks:
            print("Topper Name:", name)
            print("Highest Marks:", highest_marks)

top({"Omkar": 90, "Komal": 85, "Nikhil": 95})
```

10. Write a function that accepts a sentence and returns the frequency of each word using a dictionary.  

```python
def frequency_count(string):
    frequency = {}
    for ch in string:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1
    return frequency
str1 = "Programming"
print(frequency_count(str1))
```

11. Write a function that accepts a list of integers and returns:  • Sum of all numbers  • Count of even numbers  • Count of odd numbers 

```python
def check(a):
    even = 0
    odd = 0

    for i in a:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1

    print("Sum:", sum(a))
    print("Even:", even)
    print("Odd:", odd)

check([1, 2, 3, 4, 5])
```
