1. Write a program to count the frequency of each character in a string using a dictionary. 
-Input: "programming" -Output: {'p':1, 'r':2, 'o':1, ...} 

```python
str1 = "programming"

freq = {}

for i in str1:
    freq[i] = freq.get(i, 0) + 1    # for counting how many times an element appears in a dictionary.

print("Frequency of each character :")
print("--------------------")
print(freq)
```

2. Write a program to count the frequency of each element in a list using a dictionary. -Input : [10,20,10,30,20,10] -Output: {10:3, 20:2, 30:1} 

```python
ListOne = [10,20,10,30,20,10]

freq = {}

for i in ListOne:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print("Frequency of each element :",freq)
    
```

3. Write a program to find the key having the maximum value in a dictionary. -Input: {'A':85, 'B':92, 'C':78} Output: B 

```python
marks = {'A': 85, 'B': 92, 'C': 78}

max_key = None
max_value = 0

for i,j in marks.items():
    if j > max_value:
        max_value = j
        max_key = i

print("Maximum value in dictionary is :",max_key)

```
