1. if Statement 

```python
age = 20     # age = 10 Condition is false hence don't print anything. 

if age >= 18:
    print("eligible for voating")
```

2. if-else Statement 

```python
age = 10

if age >= 18:   # true / fale
    print("eligible for voating")

else:
    print("not eligible for voating")
```

3. if-elif-else Statement 

```python
marks = 35

if marks >= 75:   # false
    print("Grade A")

elif marks >= 55:  # false
    print("Grade B")

elif marks >= 35: #true
    print("Grade C")

else:
    print("Fail")
```

4. Match Case 

```python
grade = "b"

temp = grade.upper()

match temp:

    case "A":
        print("Marks is Above 75")

    case "B":
        print("Marks is Above 55")

    case "C":
        print("Marks is Above 35")

    case _:
        print("Fail")
```

5. Conditional Expression (Ternary Operator) 

```python
age = 10

print("eligible for vote") if age >=18 else print("not eligible")
```

6. Nested if Statement 

```python
# A Ternary Operator is a short way of writing an if-else statement in a single line. 
age = 20
nationality = "indian"

if age >= 18:

    if nationality == "indian":
        print("eligible for vote")
    else:
        print("not eligible")
else:
    print("age is not 18")
```
