1. Create a Person parent class with name and age. Create a Student child class with roll_no and marks. Use super() to initialize the parent attributes. 

```python
class Person:
    def __init__(self, name,age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, roll_no, marks):
        super().__init__(name, age)
        self.roll_no = roll_no
        self.marks = marks

    def display(self):
        print("Name is :",self.name)
        print("Age is :",self.age)
        print("Roll_no is :",self.roll_no)
        print("Marks :",self.marks)

obj1 = Student("Omkar", 22, 101, 98)

obj1.display()
```

3. Create an Employee parent class with name and salary. Create a Manager child class with an additional department attribute. Display all details using constructors. 

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def display(self):
        print("Employee name is :",self.name)
        print("Employee's salary is :",self.salary)
        print("Employee's department is :",self.department)

obj1 = Manager("Omkar", 50000, "Developer")

obj1.display()
```

7. Bank Account Create a BankAccount class with a private attribute __balance.  Create get_balance() to view the balance.  Create set_balance() to update the balance only if the amount is valid.  Create a withdraw() method with proper balance validation. 

```python
class BankAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.__balance = balance

    def get_balance(self):
        return self.__balance 

    def set_balance(self, amount):
            if amount >= 0:
                self.__balance = amount
                print("Balance update successfully.")
                
            else:
                print("You enter invalid amount.") 

    def display(self):
        print("Account number is :",self.account_no) 
        print("Your balance is :",self.__balance)
        print("--------------------------------")

class Withdraw(BankAccount):
    def __init__(self, account_no, balance):
        super().__init__(account_no, balance)

    def Withdraw(self, amount):
        if amount <= 500:
            print("Please enter amount greater than 500.")

        elif amount > self.get_balance():
            print("Your bank balance is not sufficient.")

        else:
            self.set_balance(self.get_balance() - amount)
            print("Withdraw successfully.")

        print("Remaining balance :",self.get_balance())

obj1 = Withdraw(12345, 50000)

obj1.display()
obj1.set_balance(70000)
obj1.Withdraw(10000)

```

8. Student Marks Create a Student class with private attributes __name and __marks.  Create getter and setter methods for marks.  Allow marks only between 0 and 100.  Create a child class ExamStudent that inherits from Student

```python
class Student:
    def __init__(self, name, marks):
        self.__name = name
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print("Marks updated.")

        else:
            print("Marks must be between 0 and 100")


class ExamStudent(Student):
    def display(self):
        print("Marks:", self.get_marks())


obj1 = ExamStudent("Omkar", 85)

print("Marks:", obj1.get_marks())

obj1.set_marks(95)
print("Updated Marks:", obj1.get_marks())

obj1.set_marks(120)
```

9. Constructor + Private Data Create a User class with private __password.  Initialize it through the constructor.  Create get_password() and set_password().  Setter should update the password only if it contains at least 6 characters. 

```python
class User:
    def __init__(self, password):
        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if len(self.__password) >= 5:
            self.__password = password
            print("Password is updated.")

        else:
            print("Password is invalid.")

obj1 = User("12345")

print("Password:", obj1.get_password())

obj1.set_password("python123")
print("New Password is :",obj1.get_password())

obj1.set_password("abc")
```

10. Multilevel Inheritance + Getter/Setter Person → private age with getter/setter.  Employee → salary.  Manager → department.  Use super() at each level 

```python
class Person:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age

        else:
            print("Invalid age.")

class Employee(Person):
    def __init__(self, age, salary):
        super().__init__(age)
        self.salary = salary

class Manager(Employee):
    def __init__(self,age, salary, department):
        super().__init__(age, salary)
        self.department = department

    def display(self):
        print("Age is :",self.get_age())
        print("Salary is :",self.salary)
        print("Department is :",self.department)


obj1 = Manager(22, 50000, "CS")

obj1.display()
print("!!--------After updating-------!!")
obj1.set_age(30)
obj1.display()
```
