1. while Loop 

```python
# The while loop repeatedly executes a block of code as long as the specified condition is True. 

"""while Condition:
    Block"""

i = 1

while i < 5:  # false 
    print("hello")
    i += 1
```

2. for Loop 

```python
listOne = ["Mohit","Rohit","Ram","Sham"]

for i in listOne:
    print(i)


for i in range(1,11):
    # 1 , 2 , 3 , 4 , ......, 10
    print(i)
```

```python
fruits = ["Apple","Mango","Banana"] 
for fruit in fruits: 
    print(fruit) 
```

3. range()

```python
for i in range(1,11):
    # 1 , 2 , 3 , 4,....,10
    print(i)
```

4. break Statement 

```python
for i in range(1,11):
    if i == 5: # false , false , false  , false , true
        break

    print(i)
```

5. continue Statement 

```python
# The continue statement skips the current iteration and moves to the next iteration of the loop. 

for i in range(1,11):

    if i == 5: # false , false , false  , false , true
        continue

    print(i)
```

6. pass Statement 

```python
# The pass statement is a placeholder that does nothing. It is used when a statement is syntactically required but no code needs to be executed.

for i in range(1,11):

    if i == 5: # false , false , false  , false , true
        pass

    print(i)
```

7. Nested Loops 

```python

for i in range(1,4):
    for j in range(1):
        print("*" , end=" ")   
    print("")
    j +=1
```
