1. Create a Student Management System using Class and Object in Python. 

```python
class Student:
    def __init__(self, Student_name, Roll_no, Age, Mark1, Mark2, Mark3):       # Constructor
        self.Student_name = Student_name
        self.Roll_no = Roll_no
        self.Age = Age
        self.Mark1 = Mark1
        self.Mark2 = Mark2
        self.Mark3 = Mark3
        

    def display_details(self):                             # method 1 : display_details()
        print("!!---Student_Details---!!")
        print("Name of student :",self.Student_name)
        print("Roll no :",self.Roll_no)
        print("Age of student :",self.Age)
        print("Marks of Sub1 :",self.Mark1)
        print("Marks of Sub2 :",self.Mark2)
        print("Marks of Sub3 :",self.Mark3)
        print(" ")


    def calculate_total(self):                              # method 2 : calculate_total()
        total_marks = self.Mark1 + self.Mark2 + self.Mark3
        print("Total marks obtained is :",total_marks)
        print(" ")


    def calculate_percentage(self):
        total_marks = self.Mark1 + self.Mark2 + self.Mark3  # here we can't use above total_marks. it is not access here.
        percentage = ((total_marks/300) * 100)
        print("Obtained percentage is :",percentage)
        print(" ")

    def check_result(self):                                  # Create method 3 : check_result()
        if self.Mark1 >= 35 and self.Mark2 >= 35 and self.Mark3 >= 35:
            print("Congratulations, you are pass...!")
            print(" ")

        else:
            print("Better luck next time..!")
            print(" ")


    def  update_marks(self, Subject,new_marks):              # Create method 4 : update_marks()
        if Subject == 1:
            self.Mark1 = new_marks

        elif Subject == 2:
            self.Mark2 = new_marks 

        elif Subject == 3:
            self.Mark3 = new_marks 

        else:
            print("Invalid Input.")  

    print("Marks Updated Successfully..!!")
    print(" ")


Student_name = input("Enter your name :")                 # Take input from user.
Roll_no = int(input("Enter Roll number :"))
Age = int(input("Enter your Age :"))
Mark1 = float(input("Enter marks of sub1 :"))
Mark2 = float(input("Enter marks of sub2 :"))
Mark3 = float(input("Enter marks of sub3 :"))
    

Student1 = Student(Student_name, Roll_no, Age, Mark1, Mark2, Mark3)  # Create an object of class Student.



Student1.display_details()
Student1.calculate_total()
Student1.calculate_percentage()
Student1.check_result()



Subject = int(input("Enter subject number(1, 2 or 3) :"))         # Update the marks.
new_marks = float(input("Enter new marks :"))

Student1.update_marks(Subject,new_marks)

print("Details after updating..!!")                                # After updating.
Student1.display_details()
Student1.calculate_total()
Student1.calculate_percentage()
Student1.check_result()


```

2. Create a Library Management System using Class and Object in Python. 

```python
class Book:
    def __init__(self, Book_name, Book_id, Author_name, Availability):
        self.Book_name = Book_name
        self.Book_id = Book_id
        self.Author_name = Author_name
        self.Availability = Availability

    def display_book(self):
        print("!!---Book Details---!!")
        print(" ")
        print("Name of book is :",self.Book_name)
        print("Book Id is :",self.Book_id)
        print("Author of book is :",self.Author_name)
        if self.Availability:
            print("Book is available..!")
            print(" ")

        else:
            print("Book is not available..!")
            print(" ")


    def issue_book(self):
       
        if self.Availability:
            self.Availability == False
            print("Book issued successfully..!")
            print(" ")

        else:
            print("Book is allready issued..!")
            print(" ")
    

    def return_book(self):
        if self.Availability:
            print("Book is not return until..!")
            print(" ")

        else:
            self.Availability = True
            print("Book returned successfully..!!")
            print(" ")




Book1 = Book("Shyamchi Aai", 1001, "Sane Guruji", True)
Book2 = Book("Mrutyunjay", 1002, "Shivaji Sawant", False)


Book1.display_book()
Book1.issue_book()
Book1.return_book()

Book2.display_book()
Book2.issue_book()
Book2.return_book()



        
```

3. Create a simple Employee Management System using Class and Object in Python.

4. Bank ATM.


```python
class BankATM:

    def __init__(self):
        self.pin = 0
        self.balance = 0

    def menu(self):
        print("1. Set PIN")
        print("2. Check Balance")
        print("3. Deposit")
        print("4. Withdraw")
        print("5. Change PIN")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        match choice:

            case 1:
                self.setpin()

            case 2:
                self.check_balance()

            case 3:
                self.deposit()

            case 4:
                self.withdraw()

            case 5:
                self.change_pin()

            case 6:
                print("Thank you for using ATM")

            case _:
                print("Invalid choice")
                self.menu()

    def setpin(self):

        if self.pin == 0:
            setpin = int(input("Enter PIN: "))
            self.pin = setpin
            print("PIN set successfully")

        else:
            print("PIN is already set")

        self.menu()

    def check_balance(self):

        setpin = int(input("Enter PIN: "))

        if setpin == self.pin:
            print("Bank balance:", self.balance)

        else:
            print("Invalid PIN")

        self.menu()

    def deposit(self):

        setpin = int(input("Enter PIN: "))

        if setpin == self.pin:
            amount = int(input("Enter amount to deposit: "))

            self.balance = self.balance + amount

            print("Amount deposited successfully")
            print("Bank balance:", self.balance)

        else:
            print("Invalid PIN")

        self.menu()

    def withdraw(self):

        setpin = int(input("Enter PIN: "))

        if setpin == self.pin:

            amount = int(input("Enter amount to withdraw: "))

            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Please collect your cash")
                print("Bank balance:", self.balance)

            else:
                print("Insufficient balance")

        else:
            print("Invalid PIN")

        self.menu()

    def change_pin(self):

        old_pin = int(input("Enter old PIN: "))

        if old_pin == self.pin:

            new_pin = int(input("Enter new PIN: "))
            self.pin = new_pin

            print("PIN changed successfully")

        else:
            print("Invalid PIN")

        self.menu()


obj = BankATM()
obj.menu()
```
