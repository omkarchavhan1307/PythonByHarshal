```python
# Global scope
name = "Omkar"

def one():           # Enclosing 
    print(name)

    def two():       # Local
        name = "Komal"
        print(name)

    two()

one()

    
```

```python
# 1. Built-in Namespace

# print()
# len()
# sum()
# max()
# min()
# type()
# int()
# str()
# list()
numbers = [10, 20, 30]

print(len(numbers))
```

```python
# 2. Global Namespace
# The global namespace contains names created at the top level of a Python file/module.

name = "Omkar"
age = 22

def display():
    print(name)
    print(age)

display()
```

```python
# 3. Local Namespace
# The local namespace is created when a function is called.

def student():
    name = "Omkar"
    age = 22

    print(name)
    print(age)

student()
```

```python
# 4. Enclosing Namespace
# The enclosing namespace occurs when we have nested functions.

def outer():
    name = "Omkar"

    def inner():
        print(name)

    inner()

outer()
```

```python
# Simple Example 

x = 10

def outer():
    y = 20

    def inner():
        z = 30
        print(x)
        print(y)
        print(z)

    inner()

outer()

""" Built-in Namespace
       ↓
     print()

Global Namespace
       ↓
     x = 10
     outer()

Enclosing Namespace
       ↓
     y = 20

Local Namespace
       ↓
     z = 30
     """
```

```python
# LEGB Rule
# Python uses the LEGB rule to find a name.
# Local
#   ↓
# Enclosing
#   ↓
# Global
#   ↓
# Built-in

x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        x = "Local"
        print(x)           # Built-in

    inner()

outer()
```
