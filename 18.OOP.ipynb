1. Class : A class is a blueprint or template for creating objects.

```python
class Student:
    name = "Omkar"
    age = 22
    marks = 85
```

2. Object : An object is an instance of a class.

```python
class Student:             # Create a class
    name = "Omkar"         # Define logic here
    age = 22
    marks = 90

Student1 = Student         # Create object of class here
print(Student1.name)       # Access objects and variables from class
print(Student1.age)
```

3. __init__() Constructor : The __init__() method is called automatically when an object is created.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Omkar", 22)

print(student1.name)
print(student1.age)
```

4. Methods : A function defined inside a class is called a method.

```python
class Student:

    def display(self):            # Method
        print("Hello Student")

student1 = Student()

student1.display()
```

5. Attributes : Variables associated with an object or class are called attributes.

```python
class Student:

    def __init__(self, name, age):
        self.name = name               # self.name is attribute.
        self.age = age                 # self.age is attribute.
```
