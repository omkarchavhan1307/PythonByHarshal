1. Write a program to count the frequency of each character in a string using a dictionary. 
-Input: "programming" -Output: {'p':1, 'r':2, 'o':1, ...} 

```python
str1 = "programming"

freq = {}

for i in str1:
    freq[i] = freq.get(i, 0) + 1    # for counting how many times an element appears in a dictionary.

print("Frequency of each character :")
print("--------------------")
print(freq)
```

2. Write a program to count the frequency of each element in a list using a dictionary. -Input : [10,20,10,30,20,10] -Output: {10:3, 20:2, 30:1} 

```python
ListOne = [10,20,10,30,20,10]

freq = {}

for i in ListOne:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print("Frequency of each element :",freq)
    
```

3. Write a program to find the key having the maximum value in a dictionary. -Input: {'A':85, 'B':92, 'C':78} Output: B 

```python
marks = {'A': 85, 'B': 92, 'C': 78}

max_key = None
max_value = 0

for i,j in marks.items():
    if j > max_value:
        max_value = j
        max_key = i

print("Maximum value in dictionary is :",max_key)

```

4. Write a program to find the key having the minimum value in a dictionary. 

```python
dict1 = {'A': 85, 'B': 92, 'C': 78}

min_key = ""
min_value = 100

for key in dict1:
    if dict1[key] < min_value:
        min_value = dict1[key]
        min_key = key

print("The key having minimum value :", min_key)
```

6. Write a program to remove duplicate values from a dictionary. -Input: {'a':10, 'b':20, 'c':10, 'd':30} Output: {'a':10, 'b':20, 'd':30} 

```python
dict1 = {'a':10, 'b':20, 'c':10, 'd':30} 
result = {}
seen = set() 

for i,j in dict1.items(): 
    if j not in seen: 
        result[i]=j 
        seen.add(j)
print(result)
```

8. Write a program to find the sum of all values in a dictionary. -Input: {'Math':85, 'Science':90, 'English':80} -Output: 255 

```python
dict1 = {'Math':85, 'Science':90, 'English':80} 
sum = 0
for i in dict1:
    sum = sum + dict1[i] 
print(sum) 
```

9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. -Keys = ['A', 'B', 'C'] Values = [10, 20, 30] 

```python
keys = ['A', 'B', 'C']
values = [10, 20, 30]
result = {}

# dict1 = dict(zip(keys, values))
# print(dict1)
for i in keys:
    for j in values:
     result[i] = j
print(result)
```

10. Write a program to create a student result dictionary containing name and marks of 5 students, then display the topper. 

```python
student = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    student[name] = marks

topper = ""
highest = 0

for name in student:
    if student[name] > highest:
        highest = student[name]
        topper = name

print("Topper:", topper)
print("Marks:", highest)
```
