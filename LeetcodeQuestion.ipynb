2. Find middle of two lists.If two element is their then find avg between them.

```python
listone = [1,2,3,4]
listtwo = [5,6,8,10,11]

listone.extend(listtwo)

print(listone)

if len(listone) % 2 == 1:
    middle = len(listone) // 2
    print("Middle element is :",listone[middle])

else:
    middle = len(listone) // 2
    previous = middle + 1
    median = (listone[middle] + listone[previous]) / 2    

    print("Middle number is :",median)
```

11. Majority element. If 2 elements have equal count then you have to find greatest number from them.

```python
numbers = [3,2,3,2,3,4,2,4,3,2]
majority_count = 0
largest_element = None

for i in numbers:
    frequency = numbers.count(i)

    if frequency > majority_count:
        majority_count += frequency
        largest_element = i

print("Maximum frequency :",majority_count)
print("Largest element is :", largest_element)
```
