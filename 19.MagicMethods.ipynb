```python
class Demo:
    def __init__(self, num1):
        self.numone = num1

    def __str__(self):
        return self.numone

    def __len__(self):
        return len(self.numone)

    def __eq__(self, value):
        self.numone == value.numone

obj1 = Demo(10)
obj2 = Demo(20)

# print(obj1)
# print(len(obj2))
print(obj1 == obj2)
```
