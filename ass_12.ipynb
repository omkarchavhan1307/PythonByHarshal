2. What will be the output? Explain the LEGB rule step by step. 

```python
x = "Global"           # Global scope
def outer():           # Enclosing 
    x = "Enclosing" 

    def inner():        # Local scope
        print(x)        # Built-in

    inner()  
outer() 

# O/P : Enclosing

# Local scope
#      ↓
# Enclosing scope
#      ↓
# Global Scope
#      ↓
# Built-in 
```

3. Identify the Local, Enclosing, Global, and Built-in names in the following program:  

```python
x = 10                                 # Global Scope
def outer():                           # Enclosing Scope
    y = 20  

    def inner():                       # Local Scope
        z = 30  
        print(x, y, z, len("Python"))  # Print() and Len() Are Built-in

    inner() 

outer() 

# O/P : 10 20 30 6
```

4. What will be the output? Is x being shadowed?  

```python
x = 100 
def outer(): 
    x = 200 

    def inner():
        x = 300  
        print(x)    

    inner()    
    print(x)

outer()
print(x) 

# O/P : 300
#       200
#       100

# Yes, x being shadowed at different scope levels.

# Global Scope
#     x = 100
#        ↓
# outer() : Local Scope
#     x = 200
#        ↓
# inner() : Local Scope
#     x = 300
```

6. Correct the following program using the global keyword so that the final output is 20. 

```python
x = 10 
def update():
    global x    
    x = 20  

update() 
print(x) 

### when we use global keyword it takes value of x in a function whether we print x outside to the function.
# Also when we not declare the value of x inside the function it takes the value from outside the function.
```

7. What will be the output? Explain the difference between the Local and Enclosing Namespace.  

```python
def outer(): 
    x = 10      

    def inner():   
        x = 20  
        print(x)    

    inner()     
    print(x) 

outer() 

# O/P : 20
#       10

# Local : Scope is limited for a single function. It can't access outside that function.
# Enclosing : There is Global variable is outside the function and another nested function having local scope is inside that enclosing function.
```

8. What will be the output? Why does nonlocal change the value of x?  

```python
def outer():    
    x = 10      

    def inner():        
        nonlocal x       
        x = 50    

    inner()     
    print(x) 

outer() 

# O/P : 50  

# Here we can access the local variable outside the local scope. i.e. we access the local variable in enclosing scope.
```
