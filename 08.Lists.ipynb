```python
listone = ["user1","user2"]
print("original :",listone)
print(id(listone))
print("--------------------")

copyone = listone.copy()   # shallow copy
print("copy1 :",copyone)
print(id(copyone))

print("--------------------")

copytwo = listone         # deep copy / Allising
print("copy2",copytwo)
print(id(copytwo))
copytwo.append("user3")


print("original :",listone)
print("copy1 :",copyone)
print("copy2",copytwo)
```
