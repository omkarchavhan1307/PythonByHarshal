```python
class Bank:
    def __init__(self):
        self.balance = 0


    def menu(self):
        print("1. deposite ")
        print("2. withdraw ")
        print("3. exit")

        choice = int(input("enter your choice :"))

        match choice:
            case 1:
                self.deposite()

            case 2:
                self.withdraw()

            case 3:
                self.exit()


    def deposite(self):
        ammount = int(input("Enter ammount for deposite"))

        self.balance += ammount
        print("currunt ballance is :",self.balance)
        self.menu()


    def withdraw(self):
        with_ammount = int(input("enter withdraw ammount :"))
        if self.balance >= with_ammount:
            self.balance -= with_ammount
            print("currunt balance is :",self.balance)
        else:
            print("insufficint fund")

        self.menu()

    def exit(self):
        print("thank you !!!")



obj1 = Bank()
obj1.menu()
```
