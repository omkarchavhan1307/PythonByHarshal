type conversion

1) Implicit Type Conversion (Type Casting by Python) 

```python
num_int = 10       # int 
num_float = 3.5    # float
result = num_int + num_float
print(result)      # 13.5 (float) 
print(type(result)) # <class 'float'> 
```

```python
num1 = 10         # int
num2 = 10.15      # float


res = num1 + num2

print(res)
print(type(res))
```

```python
num1 = 10
num2 = 20

temp1 = int(num1)   # 10
temp2 = float(num2)   # 20.0

res = temp1 + temp2


print(res)


user_name = "demo"
user_surname = "xyz"

full_name = user_name + user_surname

print(full_name)
```

```python
num1 = "10"
num2 = "20.10"

print(num1 + num2)

temp1 = int(num1)   # 10
temp2 = float(num2)   # 20.0
temp3 = int(temp2)

res = temp1 + temp3

print(res)
```

2. Explicit Type Conversion (Type Casting by User) 

```python
# String to int
num_str = "100" 
num_int = int(num_str) 
print(num_int + 50)   # 150  

# Float to int
pi = 3.14159 
print(int(pi))        # 3  

# Int to string 
age = 21
print("Age is " + str(age))  # Age is 21 
```

```python
percentage = 80.70
print(type(percentage))

temp = int(percentage)
print(temp)
```
