Assignment Operators

```python
# =  Assignment op
# right side value assign to left side 

user_name = "Mohit"
print(user_name)

#add and Assign
num1 = 10
# num1 = num1 + 5
num1 += 5

print(num1)


# Subtract and Assign
# num1 = num1 - 5
num1 -=5

print(num1)

#Multiply and Assign
# num1 = num1 * 5 
num1 *= 5

print(num1)

#Divide and Assign 
# num1 = num1 / 5
num1 /= 5
print(num1)

#Floor Divide and Assign
#num1 = num1 // 5
num1 //= 5
print(num1)

#Modulus and Assign
#num1 = num1 % 5
num1 %= 5
print(num1)

#Exponent and Assign
# num1 = num1 **= 5
num1 **= 5
print(num1) 

```

Arithmetic Operators  

```python
#Floor Division 
num1 = 10
num2 = 3

res = num1 // num2

print(res)
```

```python
#Modulus 
num1 = 10
num2 = 3

res = num1 %  num2

print(res)
```

Comparison Operators

```python
# Equal To

num1 = 10
num2 = 20

print(num1 == num2)            # Equal to

print(num1 != num2)            # Not Equal to


num3 = 20
num4 = 40

print( num3 > num4 )          # Greater than

print(num3 < num4)            # less than
```

```python
num1 = 15
num2 = 10

print(num1 <= num2)            # Greater Than or Equal To 
print(num1 >= num2)            # Less Than or Equal To 
```

Logical Operators

```python
user_Age = 19
nationality = "indian"

# and --> both condition true

# or --> any one condition is true
#    10     >= 10

if user_Age >= 18 or nationality == "india":

    print("eligible for voating")
else:
    print("not eligible for voating")
```

```python
age = 28

if not age >=18:                     # Not = reverse the boolean value
    print("eligible for voating")
else:
    print("not eligible for voating")
```
