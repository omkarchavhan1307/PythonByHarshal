2. Write a Python program using nested loops to find and print all prime numbers between 1 and 100.  

```python
for num in range(2,101):
    is_prime = True
    for i in range(2,num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")
```

8. Given a dictionary of students and marks, find and print the student(s) who scored the highest marks.

10. Given two dictionaries containing product names and prices, merge them into one dictionary. If a product exists in both, keep the higher price.  

```python
products1 = {"Laptop": 50000, "Mobile": 20000, "Mouse": 500}

products2 = {"Laptop": 55000, "Keyboard": 1500, "Mouse": 400}

merged = products1.copy() # {"Laptop": 50000, "Mobile": 20000, "Mouse": 500}

for product, price in products2.items():

    if product in merged:
        if price > merged[product]:
            merged[product] = price
    else:
        merged[product] = price

print(merged)
```

11. Create a function calculator(a, b, operation) that performs addition, subtraction, multiplication, and division based on the given operation.  

```python
def calculator(a, b, operation):

    if operation == "+":
        return a + b

    elif operation == "-":
        return a - b

    elif operation == "*":
        return a * b

    elif operation == "/":
        if b != 0:
            return a / b
        else:
            return "Cannot divide by zero"

    else:
        return "Invalid Operation..!"


print("Addition is :",calculator(10, 5, "+"))
print("Subtraction is :",calculator(10, 5, "-"))
print("Multiplication is :",calculator(10, 5, "*"))
print("Division is :",calculator(10, 5, "/"))
```

12. Create a function that accepts a list of numbers and returns the largest number, smallest number, sum, and average.  

```python
def calculate(numbers):

    largest = numbers[0]
    smallest = numbers[0]
    total = 0

    for i in numbers:

        if i > largest:
            largest = i

        if i < smallest:
            smallest = i

        total += i

    average = total / len(numbers)

    return largest, smallest, total, average


numbers = [10, 20, 30, 40, 50]

largest, smallest, total, average = calculate(numbers)

print("Largest :", largest)
print("Smallest :", smallest)
print("Sum :", total)
print("Average :", average)
```

13. Create a BankAccount class with account holder, account number, and balance. Add methods for deposit, withdrawal, and displaying balance. Prevent withdrawal when the balance is insufficient.  

```python
class BankAccount:

    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(" ")
        print("Amount deposited :", amount)
        print("Balance after deposit :",self.balance)
        print(" ")

    def withdraw(self, amount):

        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn :", amount)
            print("Balance after withdraw :",self.balance)
            print(" ")
        else:
            print("Insufficient balance")
            print(" ")

    def display_balance(self):
        print("Account Holder :", self.account_holder)
        print("Account Number :", self.account_number)
        print("Balance :", self.balance)


account = BankAccount("Omkar", "123456", 10000)

account.display_balance()

account.deposit(5000)

account.withdraw(3000)

account.withdraw(20000)

account.display_balance()
```

14. Create an Employee class with employee name, ID, department, and salary. Add methods to display details, increase salary by a given percentage, and check whether the salary is above ₹50,000.  

```python
class Employee:

    def __init__(self, name, emp_id, department, salary):
        self.name = name
        self.emp_id = emp_id
        self.department = department
        self.salary = salary

    def display_details(self):
        print("Name :", self.name)
        print("ID :", self.emp_id)
        print("Department :", self.department)
        print("Salary :", self.salary)
        print(" ")

    def increase_salary(self, percentage):
        increase = self.salary * percentage / 100
        self.salary += increase
        print("New Salary :", self.salary)
        print(" ")

    def check_salary(self):
        if self.salary > 50000:
            print("Salary is above ₹50,000")
        else:
            print("Salary is not above ₹50,000")


employee = Employee("Omkar", 101, "IT", 45000)

employee.display_details()

employee.increase_salary(20)

employee.check_salary()
```

15. Create a Vehicle class with brand and speed. Create Car and Bike child classes with additional attributes. Use inheritance and method overriding to display their complete details. 

```python
class Vehicle:

    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def display_details(self):
        print("Brand :", self.brand)
        print("Speed :", self.speed, "km/h")
        print(" ")


class Car(Vehicle):

    def __init__(self, brand, speed, number_of_doors):
        super().__init__(brand, speed)
        self.number_of_doors = number_of_doors

    def display_details(self):
        print("Car Details:")
        print("Brand :", self.brand)
        print("Speed :", self.speed, "km/h")
        print("Number of doors :", self.number_of_doors)
        print(" ")


class Bike(Vehicle):

    def __init__(self, brand, speed, engine_cc):
        super().__init__(brand, speed)
        self.engine_cc = engine_cc

    def display_details(self):
        print("Bike Details:")
        print("Brand:", self.brand)
        print("Speed:", self.speed, "km/h")
        print("Engine:", self.engine_cc, "CC")
        print(" ")


car = Car("Toyota", 180, 4)

bike = Bike("Yamaha", 150, 155)

car.display_details()

print()

bike.display_details()
```
