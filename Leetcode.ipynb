2. Find middle of two lists.If two element is their then find avg between them.

```python
listone = [1,2,3,4]
listtwo = [5,6,8,10]

listone.extend(listtwo)

print(listone)

if len(listone) % 2 == 1:
    middle = len(listone) // 2
    print(listone[middle])

else:
    middle = len(listone) // 2
    previous = middle + 1
    median = (listone[middle] + listone[previous]) / 2    

print("Middle number is :",median)
```
